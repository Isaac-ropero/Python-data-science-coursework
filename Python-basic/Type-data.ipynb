```python
# Print the type of 1.0

type(1.0) # Notice that 1 is an int, and 1.0 is a float
```

```python
# Print the type of 0.5

type(0.5)
```

### Converting from one object type to a different object type

You can change the type of the object in Python; this is called typecasting. 

```python
# Verify that this is an integer

type(2)
```

```python
# Convert 2 to a float

float(2)
```

```python
# Convert integer 2 to a float and check its type

type(float(2))
```

> When we convert an integer into a float, we don't really change the value (i.e., the significand) of the number. However, if we cast a float into an integer, we could potentially lose some information.

```python
# Casting 1.1 to integer will result in loss of information

int(1.1)
```

```python
# Convert a string into an integer

int('1')
```

```python
# But if you try to do so with a string that is not a perfect match for a number, you'll get an error.
# Convert a string into an integer with error

int('1 or 2 people')
```

```python
# Convert the string "1.2" into a float

float('1.2')
```

```python
# Convert an integer to a string

str(1)
```

```python
# Convert a float to a string

str(1.2)
```
